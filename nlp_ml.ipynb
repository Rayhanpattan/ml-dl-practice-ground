```python
import pandas as pd
dataset=pd.read_csv("/content/IMDB Dataset.csv")
dataset
```

```python
import re
import nltk
from nltk.stem import PorterStemmer
from nltk.corpus import stopwords
nltk.download("stopwords")
ps=PorterStemmer()
all_stopwords=set(stopwords.words("english"))
def cleaned_data (text):
  text=re.sub(r'<br\s*/?>',' ',text)
  text=re.sub('[^a-zA-z]',' ',text)
  text=text.lower()
  words=text.split()
  clean_review=[ps.stem(word) for word in words if not  word in all_stopwords ]
  return ' '.join(clean_review)
dataset['cleaned_review']=dataset["review"].apply(cleaned_data)
display(dataset[['review','cleaned_review']].head())





```
