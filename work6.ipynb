```python
import pandas as pd
import numpy as np
import sklearn
```

```python
columns = [
    'Pregnancies', 'Glucose', 'BloodPressure', 'SkinThickness', 
    'Insulin', 'BMI', 'DiabetesPedigree', 'Age', 'Outcome'
]
dataset=pd.read_csv("ndians-diabetes.csv",names=columns)
dataset
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.layers import Dropout
from tensorflow.keras.callbacks import EarlyStopping
X=dataset.drop(columns=["Outcome"])
y=dataset["Outcome"]
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=42)
scaler=StandardScaler()
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.transform(X_test)
ann=Sequential()

ann.add(Dense(units=12,activation="relu",input_shape=(X_train_scaled.shape[1],)))
ann.add(Dense(units=8,activation="relu"))
ann.add(Dropout(0.2))
ann.add(Dense(units=1,activation="sigmoid"))

ann.compile(optimizer="adam",loss="binary_crossentropy",metrics="accuracy")
early_stop=EarlyStopping(monitor='val_loss',patience=5,restore_best_weights=True)

history=ann.fit(X_train_scaled,y_train,batch_size=30,epochs=100,validation_data=(X_test_scaled,y_test))
history
```
