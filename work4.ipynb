```python
import pandas as pd
import numpy as np
import sklearn
dataset=pd.read_csv("dataR2.csv")
dataset.isna()
```

```python
x=dataset.drop(columns=["Classification"])
corr_matrix=x.corr()
corr_matrix
```

```python
new_data=dataset.drop(columns=["Insulin"])
new_data
```

```python
new_data.describe()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
X=new_data.drop(columns=["Classification"])
y=new_data["Classification"]
X_train,X_test,y_train,y_test=train_test_split(X,y, test_size=0.3, random_state=34)
d_model=DecisionTreeClassifier(max_depth=3,random_state=34)
d_model.fit(X_train,y_train)
y_pred=d_model.predict(X_test)

y1_pred=d_model.predict(X_train)


print(f"DecisionTreeClassifier accuracy(test data):{accuracy_score(y_test,y_pred)*100:.2f}%")
print(f"DecisionTreeClassifier accuracy(train data):{accuracy_score(y_train,y1_pred)*100:.2f}%")

```

```python
ran_model=RandomForestClassifier(max_depth=3,random_state=34)
ran_model.fit(X_train,y_train)
y_pred=ran_model.predict(X_test)
accuracy2=accuracy_score(y_test,y_pred)
y2_pred=ran_model.predict(X_train)
print(f"randomforestClassifier accuracy(test data):{accuracy_score(y_test,y_pred)*100:.2f}%")
print(f"randomforestClassifier accuracy(train data):{accuracy_score(y_train,y2_pred)*100:.2f}%") 
```

```python
from sklearn.model_selection import GridSearchCV
rf=RandomForestClassifier(random_state=42)
param_grid = {
    'n_estimators': [50, 100, 150], 
    'max_depth': [3, 4, 5],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4,5]
}
grid_search=GridSearchCV(estimator=rf,param_grid=param_grid,cv=5,scoring="accuracy",n_jobs=-1)
grid_search.fit(X_train,y_train)
print("best parameters found:",grid_search.best_params_)
b_model=grid_search.best_estimator_
y3_pred=grid_search.best_estimator_.predict(X_test)
y4_pred=grid_search.best_estimator_.predict(X_train)
print(f"randomforestClassifier accuracy tuned (test data):{accuracy_score(y_test,y3_pred)*100:.2f}%")
print(f"randomforestClassifier accuracy tuned (train data):{accuracy_score(y_train,y4_pred)*100:.2f}%")



```

```python
from sklearn.linear_model import LogisticRegression
log_model=LogisticRegression()
log_model.fit(X_train,y_train)
l_pred=log_model.predict(X_test)
l1_pred=log_model.predict(X_train)
print(f"logisticregressionmodel test accuracy:{accuracy_score(y_test,l_pred)*100:.2f}%")
print(f"logisticregressionmodel train accuracy:{accuracy_score(y_train,l1_pred)*100:.2f}%")
 
```
